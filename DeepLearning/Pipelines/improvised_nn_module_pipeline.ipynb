```python
import torch
import torch.nn as nn

#model class
class Model(nn.Module):
    def __init__(self, num_features):
        super().__init__()
        self.linear1 = nn.Linear(num_features, 3)
        self.relu = nn.ReLU()

        self.linear2 = nn.Linear(3,1) #this will get 3 inputs which will come from the prev outputs
        self.sigmoid = nn.Sigmoid()

    def forward(self, features):
        linear1_out = self.linear1(features)
        relu_out = self.relu(linear1_out)
        
        linear2_out = self.linear2(relu_out)
        final_out = self.sigmoid(linear2_out)
        return final_out
```

```python
# dataset
features = torch.rand(10,5)
print(features)

# create model
model = Model(features.shape[1])
print(model)

# call model
model(features)
```

```python
# see weights and biase
print(model.linear1.weight)
print(model.linear1.bias)
print("\n")
print(model.linear2.weight)
print(model.linear2.bias)
```
