improvised pipeline with Sequential container !!

```python
import torch
import torch.nn as nn

#model class
class Model(nn.Module):
    def __init__(self, num_features):
        super().__init__()

        self.network = nn.Sequential(
            nn.Linear(num_features, 3),
            nn.ReLU(),
            nn.Linear(3,1), #this will get 3 inputs which will come from the prev outputs
            nn.Sigmoid()
        )

    def forward(self, features):
        out = self.network(features)
        return out
```

```python
# dataset
features = torch.rand(10,5)
print(features)

# create model
model = Model(features.shape[1])
print(model)

# call model
model(features)
```

```python
print(model)
# see weights and biase
print(model.network[0].weight)
print(model.network[0].bias)
print("\n")
print(model.network[2].weight)
print(model.network[2].bias)
```
