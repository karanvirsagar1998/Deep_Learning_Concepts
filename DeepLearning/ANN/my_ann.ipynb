```python
# created Dataset class
class CustomDataset(Dataset):
    def __init__(self, features, labels):
        self.features = torch.tensor(features, dtype=torch.float32) #converting features and lables to tensors
        self.labels = torch.tensor(labels, dtype=torch.long)
    
    def __len__(self):
        return len(self.features)
    
    def __getitem__(self, index):
        return self.features[index], self.labels[index]
```

```python
# create train dataset
train_dataset = CustomDataset(X_train, y_train)

len(train_dataset) # get lend
train_dataset[0] #get index
```

```python
# creating test_data
test_dataset = CustomDataset(X_test, y_test)
```

```python
# create train and test loader
train_loader = DataLoader(test_dataset, batch_size=32, shuffle=True)
test_loader = DataLoader(test_dataset, batch_size=32, shuffle=False)
```

```python
# define NN class

class MyNN(nn.Module):
    def __init__(self, features):
        super.__init__()
        self.model = nn.Sequential(
            nn.Linear(features, 128),
            nn.ReLU(),
            nn.Linear(128, 64),
            nn.ReLU(),
            nn.Linear(68, 10)
        )
    
    def forward(self, x):
        return self.model(x)
```
